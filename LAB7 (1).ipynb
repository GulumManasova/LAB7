```python
import json, glob
from collections import Counter

docs = []
meta = {m["doc_id"]: m for m in json.load(open("kb_alatech/metadata.json", encoding="utf-8"))}
for path in sorted(glob.glob("kb_alatech/docs/*.md")):
    doc_id = os.path.basename(path)[:-3]
    docs.append({"doc_id": doc_id, "text": open(path, encoding="utf-8").read(), **meta[doc_id]})
questions = json.load(open("kb_alatech/questions.json", encoding="utf-8"))

print("Документов:", len(docs), "| вопросов:", len(questions))
print("По типам:", dict(Counter(d["type"] for d in docs)))
for t in ["policy", "ticket", "order", "faq"]:
    lens = [len(d["text"]) for d in docs if d["type"] == t]
    print(f"  {t:7s} средняя длина: {sum(lens)//len(lens)} символов")

print("\nТестовые вопросы:")
for q in questions:
    print(f"  {q['id']}: {q['question']}")
```

```python
import chromadb
from chromadb.utils import embedding_functions
from langchain_text_splitters import RecursiveCharacterTextSplitter

EMB_MODEL = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
ef = embedding_functions.SentenceTransformerEmbeddingFunction(model_name=EMB_MODEL)
client = chromadb.EphemeralClient()

def make_chunks(chunk_size, overlap):
    splitter = RecursiveCharacterTextSplitter(chunk_size=chunk_size, chunk_overlap=overlap,
                                              separators=["\n\n", "\n", ". ", " ", ""])
    chunks = []
    for d in docs:
        for i, piece in enumerate(splitter.split_text(d["text"])):
            chunks.append({"id": f"{d['doc_id']}#{i}", "text": piece, "doc_id": d["doc_id"]})
    return chunks

def build_index(chunks, name):
    try:
        client.delete_collection(name)
    except Exception:
        pass
    col = client.create_collection(name, embedding_function=ef, metadata={"hnsw:space": "cosine"})
    for i in range(0, len(chunks), 256):
        b = chunks[i:i + 256]
        col.add(ids=[c["id"] for c in b], documents=[c["text"] for c in b],
                metadatas=[{"doc_id": c["doc_id"]} for c in b])
    return col

def retrieve(col, question, k=5):
    r = col.query(query_texts=[question], n_results=k, include=["documents", "metadatas", "distances"])
    return [{"text": t, "doc_id": m["doc_id"], "distance": d}
            for t, m, d in zip(r["documents"][0], r["metadatas"][0], r["distances"][0])]

def is_relevant(text, q):
    return any(s in text for s in q["gold_snippets"])

def evaluate(col):
    hit1 = hit5 = mrr = 0
    for q in questions:
        ranks = [i for i, c in enumerate(retrieve(col, q["question"])) if is_relevant(c["text"], q)]
        if ranks:
            hit5 += 1; hit1 += ranks[0] == 0; mrr += 1 / (ranks[0] + 1)
    n = len(questions)
    return {"hit@1": hit1 / n, "hit@5": hit5 / n, "MRR": round(mrr / n, 3)}

print("Конвейер готов")
```

```python
import pandas as pd

CONFIGS = [(200, 30), (500, 75), (1000, 150)]   # (chunk_size, overlap)

collections, rows = {}, []
for size, ov in CONFIGS:
    chunks = make_chunks(size, ov)
    col = build_index(chunks, f"kb_{size}_{ov}")
    collections[(size, ov)] = col
    rows.append({"chunk_size": size, "overlap": ov, "чанков": len(chunks), **evaluate(col)})
    print(f"готово: chunk_size={size}, overlap={ov}")

results = pd.DataFrame(rows)
results
```

```python
# переменная BEST нигде не определялась (NameError при запуске «сверху вниз»).
# Теперь лучшая конфигурация выбирается автоматически — по максимальному MRR.
best_row = results.loc[results["MRR"].idxmax()]
BEST = (int(best_row["chunk_size"]), int(best_row["overlap"]))
print("Лучшая конфигурация по MRR:", BEST)

import os, time, json
from getpass import getpass
from groq import Groq

try:
    from google.colab import userdata
    key = userdata.get("GROQ_API_KEY")
except Exception:
    key = None

if not key:
    key = getpass("Groq API key: ")

os.environ["GROQ_API_KEY"] = key
llm = Groq(api_key=key)

LLM_MODELS = ["openai/gpt-oss-120b", "openai/gpt-oss-20b"]

SYSTEM = (
    "Ты ассистент сотрудников компании АлаТех. "
    "Отвечай только на основе контекста. "
    "Если ответа в контексте нет, напиши: «В документах нет информации». "
    "После ответа укажи doc_id источника в квадратных скобках. "
    "Отвечай кратко."
)

def generate_answer(question, contexts):
    ctx = "\n\n".join(f"[{c['doc_id']}] {c['text']}" for c in contexts)
    last_error = None

    for model in LLM_MODELS:
        for attempt in range(3):
            try:
                r = llm.chat.completions.create(
                    model=model,
                    temperature=0,
                    messages=[
                        {"role": "system", "content": SYSTEM},
                        {"role": "user", "content": f"Контекст:\n{ctx}\n\nВопрос: {question}"}
                    ]
                )
                return r.choices[0].message.content, model
            except Exception as e:
                last_error = e
                s = str(e).lower()
                if any(x in s for x in ["429", "rate", "timeout", "503", "unavailable"]) and attempt < 2:
                    time.sleep(10 * (attempt + 1))
                    continue
                break
    raise RuntimeError(last_error)

answers = []
for q in questions:
    ctx = retrieve(collections[BEST], q["question"])
    answer, model_used = generate_answer(q["question"], ctx)
    answers.append({
        "id": q["id"],
        "вопрос": q["question"],
        "найден ли нужный чанк": "да" if any(is_relevant(c["text"], q) for c in ctx) else "нет",
        "эталон": q["reference"],
        "ответ модели": answer,
        "model": model_used
    })
    time.sleep(1)

answers_df = pd.DataFrame(answers)
pd.set_option("display.max_colwidth", None)
answers_df
```

```python
from rapidfuzz import fuzz

def suggested_score(reference, answer):
    score = fuzz.token_set_ratio(reference.lower(), answer.lower())
    return 1.0 if score >= 68 else 0.5 if score >= 42 else 0.0

# Ручная проверка 10 строк: (оценка, комментарий)
MANUAL = {
 "q01": (0.5, "Чанк найден (актуальная редакция POL-HR-001, не устаревшая POL-HR-000). Ответ неполный: нет +4 дней при стаже свыше 3 лет."),
 "q02": (1.0, "Верно: за 14 календарных дней; делить можно, одна часть не менее 14 дней."),
 "q03": (0.0, "Ошибка ПОИСКА: в top-5 только «шумовые» чанки; POL-FIN-001 не найден. Модель честно ответила «нет информации»."),
 "q04": (0.5, "Верно «не более 2 дней», но «с какого момента» неверно: условие про испытательный срок не попало в контекст, модель взяла шумовую фразу про сроки."),
 "q05": (0.0, "Ошибка ПОИСКА: политика паролей вытеснена тикетами «Сброс пароля» (HD-11827). Модель ответила «нет информации»."),
 "q06": (1.0, "Верно: 1 час, security@alatech.kg, номер 777."),
 "q07": (1.0, "Верно: 5 рабочих дней."),
 "q08": (1.0, "Верно по существу вопроса (60 000 сом/год). Перенос остатка не спрашивали — авто-оценка 0,5 занижена."),
 "q09": (1.0, "Верно: Садыков Б. А. (ORD-087); приказ №78 с другим исполнителем не перепутан."),
 "q10": (1.0, "Верно по существу вопроса (до 5 000 сом без подтверждения) — авто-оценка 0,5 занижена."),
}

evaluation = answers_df.copy()
evaluation["оценка_авто"] = [suggested_score(r, a) for r, a in zip(evaluation["эталон"], evaluation["ответ модели"])]
evaluation["оценка"] = [MANUAL[i][0] for i in evaluation["id"]]
evaluation["комментарий"] = [MANUAL[i][1] for i in evaluation["id"]]

accuracy = evaluation["оценка"].sum() / len(evaluation)
accuracy_auto = evaluation["оценка_авто"].sum() / len(evaluation)
print(f"Точность (ручная проверка): {accuracy:.2f} ({evaluation['оценка'].sum()} / {len(evaluation)})")
print(f"Точность (автоэвристика):   {accuracy_auto:.2f}")
display(evaluation[["id", "найден ли нужный чанк", "оценка_авто", "оценка", "комментарий"]])
```
